```python
import pandas as pd
import numpy as np

df= pd.read_csv("../processed_data/cleaned_swiggy_vs_zomato.csv")
```

```python
df.columns.tolist()
```

```python
#Rating difference
df["rating_difference"] = df["swiggy_rating"] - df["zomato_rating"]
```

```python
#Revenue difference
df["revenue_difference"]=( df["swiggy_estimated_monthly_revenue_inr"]-df["zomato_estimated_monthly_revenue_inr"])
```

```python
#profit difference
df["profit_difference"] = (df["swiggy_estimated_net_profit_inr"]- df["zomato_estimated_net_profit_inr"])
```

```python
#Total monthly revenue
df["total_monthly_revenue"] = (df["swiggy_estimated_monthly_revenue_inr"]+ df["zomato_estimated_monthly_revenue_inr"])
```

```python
#Total_monthly profit
df["total_monthly_profit"] = (df["swiggy_estimated_net_profit_inr"]+ df["zomato_estimated_net_profit_inr"])
```

```python
#total_monthly_orders
df["total_monthly_orders"] = (df["swiggy_estimated_monthly_orders"] + df["zomato_estimated_monthly_orders"])
```

```python
#average delivery time
df["average_delivery_time"] = ( df["swiggy_avg_delivery_time_minutes"] + df["zomato_avg_delivery_time_minutes"]) / 2
```

```python
df.to_csv("../processed_data/swiggy_zomato_featured.csv", index=False)
```
